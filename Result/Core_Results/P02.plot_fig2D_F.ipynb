```python
import matplotlib.pyplot as plt

# 数据
categories = ['34AA', '65AA','88AA']
values1 = [6,47,39]
values2 = [28,18,49]
```

```python

# 绘图
# 设置柱子的宽度
bar_width = 0.6

plt.figure(figsize=(2.5, 3.5))

plt.bar(categories, values1, label='non-conservation',width=bar_width)
plt.bar(categories, values2, bottom=values1, label='conservation',width=bar_width)

#plt.xlabel('Predictor')
plt.ylabel('Number')
#plt.title('Stacked Bar Chart')
plt.legend()
plt.savefig("Fig2_HLA_pseudo_sequence_boxplot.svg")
plt.show()
```

```python
import matplotlib.pyplot as plt

# 数据
labels = ['HLA α1', 'HLA α2','HLA α3','other']
sizes = [23,11,14,17]
colors = [ '#EE6E33', '#F2A402','#A1BD42','#30C099']
explode = (0.1, 0,0,0)  # 突出显示第二个部分

# 绘图
plt.figure(figsize=(4, 4))

plt.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)

plt.axis('equal')  # 保持饼图为正圆形
plt.title('65AA')
plt.savefig("Fig2_HLA_pseudo_sequence_pie_LKMHCpan2nlp.svg")
plt.show()
```

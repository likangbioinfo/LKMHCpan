```python
from scipy.stats import hypergeom
from scipy.stats import fisher_exact
```

```python
# sites count
total=395
overlap=11
netMHCpan=34
LKMHCpan=65
```

```python
####LKMHCpan在netMHCpan位点富集
p_value = hypergeom.sf(overlap-1, total, netMHCpan, LKMHCpan)
print(p_value)
```

```python
# conservation sites： 相对netMHCpan，LKMHCpan非保守位占比高
netMHCpan_con=28
netMHCpan_non=6
LKMHCpan_con =18
LKMHCpan_non =47
table=[[netMHCpan_con,netMHCpan_non],[LKMHCpan_con,LKMHCpan_non]]

oddsratio, p_value = fisher_exact(table)
print(f"Odd's Ratio: {oddsratio}")
print(f"P-value: {p_value}")
```

```python
# LKMHCpan conservation sites
total_con=153
netMHCpan_con=28
LKMHCpan_con =18
overlap_con  =7

p_value = hypergeom.sf(overlap_con-1, total_con, netMHCpan_con, LKMHCpan_con)
print(p_value)
```

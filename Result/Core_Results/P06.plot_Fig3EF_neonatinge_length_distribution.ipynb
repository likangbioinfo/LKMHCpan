```python
import os
import sys
import pandas as pd
import matplotlib.pyplot as plt
```

黑色素瘤/肺癌的滑窗肽段预测为阳性的肽段长度分布

```python
def plot_bar(dict1,sample,outplot):
    # 获取所有键（x 轴标签）
    keys = list(dict1.keys())
    # 获取每组字典的值
    values1 = list(dict1.values())
    plt.figure(figsize=(3, 3))
    # 设置条形图的宽度
    #bar_width = 0.45

    # 设置 x 轴的位置
    index = range(len(keys))

    # 创建条形图
    plt.bar(index, values1, label='peptide', alpha=0.8)

    # 添加 x 轴标签
    plt.xlabel('Sample')
    plt.ylabel('peptide number')
    plt.title(sample)
    # 添加图例
    plt.legend()
    # 显示图形
    plt.tight_layout()  # 自动调整子图参数，使之填充整个图像区域
    plt.savefig(outplot)
    #plt.show()
```

```python
inputfile="Table_mutant_merges_peptide_all_NetMHCpan_predict.txt"
```

```python
df_raw=pd.read_csv(inputfile,sep="\t")
df_raw['ID'] = df_raw['ID'].astype(str)
df_fit=df_raw[df_raw['LKMHCpan2_pre']>=0.5]
```

```python
sample_list=df_fit['ID'].unique().tolist()
dict_splid,dict_pos={},{}
for sample in sample_list:
    #print(sample)
    list_splid,list_pos={},{}
    df_aim=df_fit[df_fit['ID']==sample]
    for length in range(8,15):
        df_len=df_aim[df_aim['Length']==length]
        number=df_len['Length'].value_counts().to_list()[0]
        dict_splid[length]=number

        df_len_pos=df_len[df_len['LKMHCpan2_pre']>=0.5]
        number_pos=df_len_pos['Length'].value_counts().to_list()[0]
        dict_pos[length]=number_pos
    outplot="./Length_distribut/Plot_"+sample+"_length_distribution.svg"
    plot_bar(dict_pos,sample,outplot)
```
